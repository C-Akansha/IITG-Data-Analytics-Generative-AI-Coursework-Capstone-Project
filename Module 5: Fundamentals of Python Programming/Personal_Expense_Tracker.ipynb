```python
#Step 1 Add expense data

expenses = []          #List of the expense details 
monthly_budget = None  #Stores budget values

def add_expense():
    print("Add Expense")
    date = input("Enter date (YYYY-MM-DD): ")
    category = input("Enter category : ")
    amount = float(input("Enter amount: "))
    description = input("Enter description: ")
    expense = {
        "date": date,
        "category": category,
        "amount": amount,
        "description": description
    }
    expenses.append(expense)
    print("Expense added successfully.")

add_expense()
```

```python
#Step 2 View Expenses 
def view_expenses():
    print("Expense records:")
    for exp in expenses:
        if exp['date'] == "":
            print("Skipping - missing date")
            continue
        elif exp['category'] == "":
            print("Skipping - missing category")
            continue
        elif exp['amount'] == "":
            print("Skipping - missing amount")
            continue
        elif exp['description'] == "":
            print("Skipping - missing description")
            continue
        else:
            print(exp)

view_expenses()
```

```python
#Step 3 Setting and tracking the budget

#Setting the monthly budget
def set_budget():
    budget = float(input("Enter monthly budget: "))
    print("Monthly Budget set to",budget)
    return budget

#Tracking the Budget

def track_budget():
    total = sum(exp['amount'] for exp in expenses)
    remaining = monthly_budget - total
    print("Total spent: ",total)
    if total > monthly_budget:
        print("You have exceeded your budget")
    else:
        print("You have",remaining,"left for the month")

monthly_budget = set_budget()
track_budget()
```

```python
#Step 5 To create interactive menu 

def show_menu():
    print("Personal Expense Tracker Menu")
    print("1. Add expense")
    print("2. View expenses")
    print("3. Track budget")
    print("4. Save expenses")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        add_expense()
    elif choice == "2":
        view_expenses()
    elif choice == "3":
        track_budget()
    elif choice == "4":
        print("Expense file saved.")   
    elif choice == "5":
        print("File saved and exiting program.")
    else:
        print("Invalid choice. Please enter a number from 1 to 5.")

show_menu()
```
